```python
import pandas as pd
import os
from req_osrm import get_osrm_roundtrip_edges
```

```python
list_df = [pd.read_csv(f'data-used/{v}') for v in os.listdir('data-used')]
```

```python
col_used = ['node_id','nama_node','latitude','longitude']
```

```python
df = pd.concat(list_df)[col_used]
```

```python
df = df.loc[~df['nama_node'].duplicated()].copy()
```

```python
df_result = get_osrm_roundtrip_edges(df=df)
```

```python
df_result['conn_id'] = df_result['from_node'] +' - '+ df_result['to_node']
```

```python
df_result.set_index('conn_id').to_csv('result.csv')
```
