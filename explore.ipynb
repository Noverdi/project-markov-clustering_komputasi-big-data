```python
import pandas as pd
```

```python
filename = 'data-used/01_node_kantor_pemerintahan_kab_bekasi_gmaps.csv'
```

```python
df = pd.read_csv(filename)
```

```python
def format_id(id):
    res  = f'PEM_{id:03}'
    return res
```

```python
df['node_id'] = df['node_id'].apply(lambda x: format_id(x))
```

```python
df.set_index('node_id').to_csv(filename)
```
